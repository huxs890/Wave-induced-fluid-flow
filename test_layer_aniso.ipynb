```python
import numpy as np
import matplotlib.pyplot as plt
plt.rcParams.update({
    'font.family': 'Times New Roman',
    'mathtext.fontset': 'stix',
    'font.size': 12,
    'axes.labelsize': 14,
    'xtick.labelsize': 12,
    'ytick.labelsize': 12,
    'legend.fontsize': 11
})

from utils import phase_velocity_vti
from wiff_layer_aniso import biot_layer_aniso,biot_aniso_background_layer
```

```python
Ks = 38e9
Gs = 44e9
rhos = 2.65e3
Kd = 10e9
Gd = 8e9
phi = 0.1
Kf_up = 2.4e9
rhof_up = 1e3
vis_up = 1e-3
Kf_low = 9.6e6
rhof_low = 50
vis_low = 1e-5
perm0 = 100e-15
d1 = 0.2
d2 = 0.2
Pc = 15e3
strain0 = 1e-5
f = np.logspace(0,5,100)

rock_up   = [Ks,Gs,rhos,Kd,Gd,phi]
fluid_up  = [Kf_up,rhof_up,vis_up]
rock_low  = [Ks,Gs,rhos,Kd,Gd,phi]
fluid_low = [Kf_low,rhof_low,vis_low]
perm      = [perm0,perm0]

C11,C13,C33,C44,C66,h,pf_C33,pf_C11 = biot_layer_aniso(rock_up,rock_low,fluid_up,fluid_low,perm,d1,d2,f,Pc,strain0)

plt.figure(figsize=(16,6))
plt.subplot(2,3,1)
plt.semilogx(f,np.real(C11)/1e9,color='blue',linewidth=3)
plt.xlabel('Frequency (Hz)')
plt.ylabel('C11 (Gpa)')
plt.xlim(f.min(),f.max())
plt.title('C11')

plt.subplot(2,3,2)
plt.semilogx(f,np.real(C13)/1e9,color='blue',linewidth=3)
plt.xlabel('Frequency (Hz)')
plt.ylabel('C13 (Gpa)')
plt.xlim(f.min(),f.max())
plt.title('C13')

plt.subplot(2,3,3)
plt.semilogx(f,np.real(C33)/1e9,color='blue',linewidth=3)
plt.xlabel('Frequency (Hz)')
plt.ylabel('C33 (Gpa)')
plt.xlim(f.min(),f.max())
plt.title('C33')

plt.subplot(2,3,4)
plt.semilogx(f,np.real(C44)/1e9,color='blue',linewidth=3)
plt.xlabel('Frequency (Hz)')
plt.ylabel('C44 (Gpa)')
plt.xlim(f.min(),f.max())
plt.title('C44')

plt.subplot(2,3,5)
plt.semilogx(f,np.real(C66)/1e9,color='blue',linewidth=3)
plt.xlabel('Frequency (Hz)')
plt.ylabel('C66 (Gpa)')
plt.xlim(f.min(),f.max())
plt.title('C66')

plt.tight_layout()
plt.show()

plt.figure(figsize=(6,4))
plt.semilogx(f,np.real(C11)/1e9,label='C11',linewidth=3,color='blue')
plt.semilogx(f,np.real(C13)/1e9,label='C13',linewidth=3,color='red')
plt.semilogx(f,np.real(C33)/1e9,label='C33',linewidth=3,color='black',linestyle='--')
plt.semilogx(f,np.real(C44)/1e9,label='C44',linewidth=3,color='Green')
plt.semilogx(f,np.real(C66)/1e9,label='C66',linewidth=3,color='blue',linestyle='--')
plt.xlabel('Frequency (Hz)')
plt.ylabel('Modulus (Gpa)')
plt.legend()
plt.xlim(f.min(),f.max())
plt.show()
```
